```python
import random

alphabet_num = 10
length = 20 # 長さは固定

alphabet = tuple(chr(i + 65) for i in range(alphabet_num))

# ルールは幾何分布を使ってランダムに生成
rule_num = 100
rule_length_expect_inv = 0.5 #幾何分布のパラメータpにあたる

rule = []
while len(rule) < rule_num:
    i = 1
    while (random.random() > rule_length_expect_inv) and (i < length):
        i += 1
    before = ''.join(random.choices(alphabet, k=i))
    after = ''.join(random.choices(alphabet, k=i))
    if (before != after) and ((before, after) not in rule): # これは必要か?
        rule.append((before, after))

print(rule)
```

```python
while True:
  before = ''.join(random.choices(alphabet, k=length))
  after = ''.join(random.choices(alphabet, k=length))
  if before != after:
    break

print(before)
print(after)
# ruleを使って到達可能か?
```

```python
import heapq

def distance(string1, string2):
    assert len(string1) == len(string2)
    return sum(c1 != c2 for c1, c2 in zip(string1, string2))

def best_first_search(before, after, rule):
    searched_set = set([before])
    candidate_queue = []
    heapq.heappush(candidate_queue, (distance(before, after), [before]))

    while candidate_queue:
        _, search_history = heapq.heappop(candidate_queue)
        current_string = search_history[-1]
        length = len(current_string)  # 修正点1

        for rule_before, rule_after in rule:
            i = len(rule_before)
            for j in range(length - i + 1):
                if current_string[j:j+i] == rule_before:
                    new_string = current_string[:j] + rule_after + current_string[j+i:]
                    if new_string in searched_set:
                        continue
                    searched_set.add(new_string)
                    new_history = search_history + [new_string]
                    heapq.heappush(candidate_queue, (distance(new_string, after), new_history))  # 修正点2

                    if new_string == after:
                        return new_history

    print("到達不可")
    return None




```

```python
path = best_first_search(before, after, rule)
if path:
    print("変換系列")
    for step in path:
        print(step)
```
