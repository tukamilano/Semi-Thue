```python
import random

alphabet_num = 10
length = 20 # 長さは固定
convert_length = 3

alphabet = tuple(chr(i + 65) for i in range(alphabet_num))

rule_num = 200

rules = []
while len(rules) < rule_num:
    before = ''.join(random.choices(alphabet, k=convert_length))
    after = ''.join(random.choices(alphabet, k=convert_length))
    if (before != after) and ((before, after) not in rules):
        rules.append((before, after))

print(rules)
```

```python
def hamming(a, b):
    assert len(a) == len(b)
    return sum(c1 != c2 for c1,c2 in zip(a,b))

def find_successors(s, rules):
    out = []
    for lhs, rhs in rules:
        k = len(lhs)
        start = 0
        while True:
            idx = s.find(lhs, start)
            if idx == -1:
                break
            nxt = s[:idx] + rhs + s[idx + k:]
            out.append((nxt, (idx, lhs, rhs)))
            start = idx + 1
    return out
```

```python
import sys

def beam_search(before, after, width=100):
    state_list = [(before, hamming(before, after))]
    while len(state_list) != 0:
        new_state_list = []
        successors_list = []
        for state in state_list:
            successors_list += find_successors(state[0], rules)
        for successor in successors_list:
            s = (successor[0], hamming(successor[0], after))
            if s not in new_state_list:
                new_state_list.append(s)
            if hamming(successor[0], after) == 0:
                print('found a path')
                sys.exit()
        state_list = sorted(new_state_list, key=lambda x: x[1])[:width]
        print(state_list)

beam_search(before,after)





```
