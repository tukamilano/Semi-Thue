```python
import random

alphabet_num = 10
length = 20
convert_length = 3

alphabet = tuple(chr(i + 65) for i in range(alphabet_num))

rule_num = 200

rules = []
while len(rules) < rule_num:
    before = ''.join(random.choices(alphabet, k=convert_length))
    after = ''.join(random.choices(alphabet, k=convert_length))
    if (before != after) and ((before, after) not in rules):
        rules.append((before, after))

print(rules)
```

やりたいこと
- 子孫の数(の分布)を数える
- 積み重ね方(の分布)を数える

```python
from collections import deque

visited = set()
queue = deque(before)

while queue:
    current = queue.popleft()
    if current in visited:
        continue
    visited.add(current)
    for i in range(len(current) - convert_length + 1):
        part = current[i:i + convert_length]
        for b, a in rules:
            if part == b:
                new_string = current[:i] + a + current[i + convert_length:]
                if new_string not in visited:
                    queue.append(new_string)
print(len(visited))


```
