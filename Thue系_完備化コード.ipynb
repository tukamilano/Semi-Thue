```python
import random

alphabet_num = 25

alphabet = tuple(chr(i + 65) for i in range(alphabet_num))

# ルールは幾何分布を使ってランダムに生成
rule_num = 100
rule_length_expect_inv = 0.2 #幾何分布のパラメータpにあたる

rule = []
while len(rule) < rule_num:
    i = 0
    while random.random() > rule_length_expect_inv:
        i += 1
    j = 0
    while random.random() > rule_length_expect_inv:
        j += 1
    before = ''.join(random.choices(alphabet, k=i))
    after = ''.join(random.choices(alphabet, k=j))
    strings = [before, after]
    strings.sort(key=lambda s: (len(s), s), reverse=True)
    rule.append(tuple(strings))

rule = sorted(rule, key=lambda k: len(k[0]), reverse=True)
print(rule)
```

```python
def run_rules(s,rules):
  old_s = None
  while s != old_s:
    old_s = s
    for lhs,rhs in rules:
      s = s.replace(lhs,rhs)
  return s

def reduce_rules(E):
  # worklist style
  R = set()
  # Sort E so smallest last probably. Most reduction power.
  E = sorted(E, key=lambda k: len(k[0]), reverse=True)
  while len(E) > 0:
    (a,b) = E.pop()
    a = run_rules(a, R) # simplify
    b = run_rules(b, R)
    if a == b: #delete
      continue
    if (len(b), b) > (len(a), a): # len then lex ordering
      R.add((b,a))
    if (len(a), a) > (len(b), b):
      R.add((a,b))
  return R
print(rule)
print(reduce_rules(rule))
```
